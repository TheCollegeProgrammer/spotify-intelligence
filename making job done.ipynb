```python
import pandas as pd

def get_yearly_song_streams(df):
    """
    Returns yearly total streams per song
    by summing monthly streams across all months and countries.
    """

    yearly_song_streams = (
        df.groupby(["track", "artist"])["monthly_streams"]
        .sum()
        .reset_index()
        .rename(columns={"monthly_streams": "yearly_streams"})
        .sort_values("yearly_streams", ascending=False)
    )

    return yearly_song_streams
```

```python
df = pd.read_csv("top5_countries_monthly_2025_clean.csv")

yearly_songs = get_yearly_song_streams(df)
yearly_song_frame = pd.DataFrame(yearly_songs, columns=['Track', 'Artist', 'Yearly_Stream'])

print(yearly_songs.head(10))

```

## Highest Stream Song Country Wise

### Subtask:
Calculate the total streams for each song within each country by grouping the original DataFrame (`df`) by 'country', 'track', and 'artist', and summing 'monthly_streams'. Then, for each country, identify the song with the highest total streams.


**Reasoning**:
First, I need to group the original DataFrame `df` by 'country', 'track', and 'artist' and sum the 'monthly_streams' to get the total streams for each song within each country, as per the first instruction.



```python
country_song_streams = df.groupby(['country', 'track', 'artist'])['monthly_streams'].sum().reset_index()
country_song_streams = country_song_streams.rename(columns={'monthly_streams': 'total_country_streams'})
print(country_song_streams.head())
```

**Reasoning**:
The next step is to identify the song with the highest total streams for each country using the `country_song_streams` DataFrame, as per the subtask instruction.



```python
highest_stream_song_per_country = country_song_streams.loc[country_song_streams.groupby('country')['total_country_streams'].idxmax()].reset_index(drop=True)
print(highest_stream_song_per_country)
```

## Plot Highest Stream Song Country Wise

### Subtask:
Generate a bar chart to visualize the highest stream song for each country. Ensure the plot has a clear title, axis labels, and a legend.


**Reasoning**:
The subtask requires generating a bar chart to visualize the highest streamed song for each country. To achieve this, I need to import the `matplotlib.pyplot` library, create a bar plot using the `highest_stream_song_per_country` DataFrame, and then customize the plot with appropriate labels, titles, and layout adjustments.



```python
import matplotlib.pyplot as plt

# Create a combined label for country, track, and artist for better readability on the x-axis
highest_stream_song_per_country['display_label'] = highest_stream_song_per_country['country'] + '\n' + \
                                                  highest_stream_song_per_country['track'] + '\n by \n' + \
                                                  highest_stream_song_per_country['artist']

plt.figure(figsize=(12, 7))
plt.bar(highest_stream_song_per_country['display_label'], highest_stream_song_per_country['total_country_streams'], color='skyblue')

plt.xlabel('Country, Track, and Artist')
plt.ylabel('Total Streams')
plt.title('Highest Streamed Song Per Country')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Highest Song Popularity (Max Growth Rate)

### Subtask:
To determine song popularity by maximum growth rate, first sort the original DataFrame (`df`) by 'track', 'artist', and 'month'. Then, for each song, calculate the monthly stream difference (current month streams - previous month streams). Identify the song and month with the maximum positive stream difference.


**Reasoning**:
To enable proper chronological sorting and calculation of monthly differences, I will convert the 'month' column in the DataFrame `df` to a datetime object.



```python
df['month'] = pd.to_datetime(df['month'])
print("Month column converted to datetime.")
```

**Reasoning**:
To prepare for calculating monthly stream differences, I need to sort the DataFrame by 'track', 'artist', and 'month' to ensure that monthly data for each song is in chronological order.



```python
df_sorted = df.sort_values(by=['track', 'artist', 'month']).reset_index(drop=True)
print("DataFrame sorted by track, artist, and month.")
```

**Reasoning**:
To calculate the monthly stream difference for each song, I need to group the `df_sorted` DataFrame by 'track' and 'artist' and then compute the difference between consecutive 'monthly_streams' values. This difference will be stored in a new column named 'stream_difference'.



```python
df_sorted['stream_difference'] = df_sorted.groupby(['track', 'artist'])['monthly_streams'].diff()
print("Monthly stream difference calculated and added to 'stream_difference' column.")
```

**Reasoning**:
To identify the song and month with the maximum positive stream difference, I will find the row in the `df_sorted` DataFrame where 'stream_difference' is highest and positive, and store it in `highest_growth_song`.



```python
highest_growth_song = df_sorted[df_sorted['stream_difference'] > 0].loc[df_sorted['stream_difference'].idxmax()]
print("Song with the highest monthly stream growth rate:")
print(highest_growth_song)
```

## Plot Highest Song Popularity (Max Growth Rate)

### Subtask:
Generate a line plot showing the monthly streams for the song identified as having the highest growth rate, to visualize its stream progression over the year. Ensure the plot has a clear title, axis labels, and a legend.


**Reasoning**:
To visualize the monthly stream progression for the song with the highest growth rate, I need to filter the `df_sorted` DataFrame to include only the data for that specific track and artist, then plot its 'monthly_streams' against the 'month'.



```python
import matplotlib.pyplot as plt

# Filter df_sorted for the highest growth song
highest_growth_track = highest_growth_song['track']
highest_growth_artist = highest_growth_song['artist']

song_data = df_sorted[(df_sorted['track'] == highest_growth_track) &
                      (df_sorted['artist'] == highest_growth_artist)].copy()

plt.figure(figsize=(12, 7))
plt.plot(song_data['month'], song_data['monthly_streams'], marker='o', linestyle='-', color='purple', label=f'{highest_growth_track} by {highest_growth_artist}')

plt.xlabel('Month')
plt.ylabel('Monthly Streams')
plt.title(f'Monthly Streams for {highest_growth_track} by {highest_growth_artist}')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
