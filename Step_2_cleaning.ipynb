```python
import pandas as pd

df = pd.read_csv("top5_countries_monthly_2023.csv")

top5_global_songs = (
    df.groupby(["track", "artist"])["monthly_streams"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print(top5_global_songs)
```

```python
df["track"] = (
    df["track"]
    .str.encode("latin1", errors="ignore")
    .str.decode("utf-8", errors="ignore")
)

df["artist"] = (
    df["artist"]
    .str.encode("latin1", errors="ignore")
    .str.decode("utf-8", errors="ignore")
)
```

```python
df.to_csv(
    "top5_countries_monthly_2023_clean.csv",
    index=False,
    encoding="utf-8"
)
```
