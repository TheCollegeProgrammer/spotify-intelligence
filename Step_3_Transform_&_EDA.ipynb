```python
import pandas as pd

def get_yearly_song_streams(df):
    """
    Returns yearly total streams per song
    by summing monthly streams across all months and countries.
    """

    yearly_song_streams = (
        df.groupby(["track", "artist"])["monthly_streams"]
        .sum()
        .reset_index()
        .rename(columns={"monthly_streams": "yearly_streams"})
        .sort_values("yearly_streams", ascending=False)
    )

    return yearly_song_streams
```

```python
df = pd.read_csv("top5_countries_monthly_2025_clean.csv")

yearly_songs = get_yearly_song_streams(df)
yearly_song_frame = pd.DataFrame(yearly_songs, columns=['Track', 'Artist', 'Yearly_Stream'])

print(yearly_songs.head(10))

```

```python
top5_country_songs = (
    df.groupby(["country", "track", "artist"])["monthly_streams"]
    .sum()
    .reset_index()
)

top5_country_songs = (
    top5_country_songs
    .sort_values(["country", "monthly_streams"], ascending=[True, False])
    .groupby("country")
    .head(5)
)

print(top5_country_songs)
```

```python
most_active_country = (
    df.groupby("country")["monthly_streams"]
    .sum()
    .sort_values(ascending=False)
)

print(most_active_country)
```

```python
top_artists = (
    df.groupby("artist")["monthly_streams"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print(top_artists)
```

```python
import pandas as pd

df = pd.read_csv("top5_countries_monthly_2025_clean.csv")

# 1️⃣ Aggregate globally per song per month
song_monthly = (
    df.groupby(["track", "month"], as_index=False)["monthly_streams"]
    .sum()
)

# 2️⃣ Keep songs appearing in at least 3 months
valid_songs = (
    song_monthly.groupby("track")["month"]
    .nunique()
    .reset_index()
)

valid_songs = valid_songs[valid_songs["month"] >= 3]["track"]

song_monthly = song_monthly[song_monthly["track"].isin(valid_songs)]

# 3️⃣ Compute growth using min → max logic
growth = (
    song_monthly
    .groupby("track", as_index=False)["monthly_streams"]
    .agg(min_streams="min", max_streams="max")
)

growth["growth_percent"] = (
    (growth["max_streams"] - growth["min_streams"])
    / growth["min_streams"] * 100
)

# 4️⃣ Top 5 fastest growing songs
top_growth = (
    growth
    .sort_values("growth_percent", ascending=False)
    .head(5)
)

print(top_growth)
```
